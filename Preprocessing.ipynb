```python
import matplotlib.pyplot as plt
import numpy as np
import rasterio
```

```python
data_path = r"Data\Processed Datasets\Dataset_V1\dataset_V1.tif" 

with rasterio.open(data_path) as src:
    ds = src.read()
    profile = src.profile

print(ds.shape)
```

```python
# show an image of the dataset in rgb

rgb = ds[:3, :, :]
rgb = rgb / np.max(rgb)
rgb = rgb.transpose(1, 2, 0)  # change shape to (rows, cols, bands)

plt.imshow(rgb)
plt.show()
```
